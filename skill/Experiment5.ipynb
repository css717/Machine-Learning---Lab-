```python
import pandas as pd
import numpy as np

# Create sample placement data matching the features in the code
data = {
    'Gender': np.random.choice(['Male', 'Female'], 50),
    'City': np.random.choice(['CityA', 'CityB'], 50),
    'CollegeTier': np.random.choice([1, 2, 3], 50),
    'Stream': np.random.choice(['CS', 'ECE', 'ME'], 50),
    'Specialisation': np.random.choice(['AI', 'Web', 'Core'], 50),
    'Hostel': np.random.choice(['Yes', 'No'], 50),
    'HistoryOfBacklogs': np.random.choice(['Yes', 'No'], 50),
    'ExtraCurricular': np.random.choice(['Yes', 'No'], 50),
    'CGPA': np.random.uniform(6.0, 9.5, 50),
    'Placed': np.random.choice([0, 1], 50)
}

df_sample = pd.DataFrame(data)
df_sample.to_csv('simple_placement_dataset.csv', index=False)
print("Created simple_placement_dataset.csv successfully!")
```
